## One-time setup

Do this once, today. I have already created your repository for you.


That last line matters more than it looks. `nbstripout` strips output cells from
notebooks before they are committed, so your diffs show what you actually
*changed* rather than the fact that you re-ran a cell and a plot came out with
different pixel values. Without it, notebook merge conflicts are miserable.

Check that both remotes are there:

```bash
git remote -v
```

You should see `origin` pointing at your repo and `upstream` pointing at `FALL26`.

# What Is Machine Learning?

> Machine learning is the science of programming computers so they can learn from data.
>
> Aurelien Geron, chapter 1

Compare two ways of writing a spam filter.

**The way you already know how to do it.** You look at spam, notice that it says
things like "free money" and comes from strange addresses, and you write rules. It
works for a while. Then spammers write "fr33 m0ney" and you write more rules.
Forever. The program only knows what you personally thought to tell it.

**The machine learning way.** You collect thousands of emails already labeled spam
or not spam, and you write a program that finds the patterns itself. When spammers
change tactics you do not rewrite the rules; you retrain on newer email.

The second program is not smarter than you. It is doing something you could do by
hand, only across far more examples than you have patience for, and it keeps doing
it after you stop paying attention.

## The kinds of learning

| Kind | What the data looks like | Example |
|---|---|---|
| **Supervised** | Examples with correct answers attached | Photos labeled cat or dog |
| **Unsupervised** | Examples with no answers | Customers, grouped by whatever structure exists |
| **Self-supervised** | Answers generated from the data itself | Hide a word in a sentence, predict the hidden word |
| **Reinforcement** | No answers, just rewards and penalties | An agent learning to play a game |

Most of this course is supervised learning. But keep **self-supervised** in the
back of your mind, because it is the trick that makes large language models
possible, and we get there in November.

## What this course is not

Instead we start from a harder and more useful question: **how do you know whether

If you have *not* taken DATA 201, you are fine. We start from regression, which you

## The cycle that runs the entire course

Almost everything we do fits this loop:

```{mermaid}
flowchart LR
  A[Frame the question] --> B[Split the data]
  B --> C[Train on the training set]
  C --> D[Tune on the validation set]
  D --> C
  D --> E[Test once on the test set]
  E --> F[Report honestly]
```

The single most important idea in this course lives in that diagram, and it is
this: **the test set is touched once.** Every time you peek at test performance and
then change something, you leak a little information about the test set into your
model, and your reported accuracy becomes a little more of a lie.

This is not a technicality. It is the difference between a model that works and a
model that only appeared to work in a notebook.

## Where large language models fit

By the end of the semester you will build a small language model from scratch and
train it on your own laptop. Not a toy demonstration of somebody else's model:

It will be small and it will not be very good. That is fine. Once you have built
