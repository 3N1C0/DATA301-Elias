Add noise, plot it, and say in one sentence what the true relationship is.

````python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=67)

m = 150 # num of data points 
X = 10 * rng.random((m,1)) - 5 # x between -5 and 5
y = 1.5 * np.sin(4*X)+4  # a sine wave plus noise

plt.figure(figsize=(6, 6))
plt.plot(X, y, "b.")
plt.xlabel("$x$"); plt.ylabel("$y$", rotation=0)
plt.axis([-3, 3, 0, 7]); plt.grid()
plt.show()
````
